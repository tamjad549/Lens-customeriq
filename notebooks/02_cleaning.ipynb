```python
import pandas as pd

df = pd.read_csv("../data/raw/online_retail_II.csv")   # apni file ka path/naam
print("Shuru me rows:", len(df))
```

```python
# Step 1: Customer ID missing wali rows hatao
# (bina ID ke pata nahi chalta ke customer kaun hai)
df = df.dropna(subset=["Customer ID"])
print("Missing ID hatane ke baad:", len(df))
```

```python
# Step 2: Cancelled orders hatao (Invoice C se shuru hota hai)
df = df[~df["Invoice"].astype(str).str.startswith("C")]
print("Cancelled hatane ke baad:", len(df))
```

```python
# Step 3: Returns aur galat values hatao (Quantity aur Price zero ya minus nahi hona chahiye)
df = df[(df["Quantity"] > 0) & (df["Price"] > 0)]
print("Returns/galat values hatane ke baad:", len(df))
```

```python
# Step 4: Duplicate rows hatao
df = df.drop_duplicates()
print("Duplicate hatane ke baad:", len(df))
```

```python
# Step 5: Types theek karo aur TotalPrice banao (ye aage "Monetary" banane me kaam aayega)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Customer ID"] = df["Customer ID"].astype(int)
df["TotalPrice"] = df["Quantity"] * df["Price"]
df.describe()
```

```python
# Step 6: Extreme outliers hatao (sabse upar ke 0.1% ke bhi upar wali values)
q_limit = df["Quantity"].quantile(0.999)
p_limit = df["Price"].quantile(0.999)
df = df[(df["Quantity"] <= q_limit) & (df["Price"] <= p_limit)]
print("Outliers hatane ke baad:", len(df))
```

```python
# Saaf data save karo (agla notebook isi se shuru hoga)
df.to_csv("../data/processed/clean_retail.csv", index=False)
print("Saved:", df.shape)
```
