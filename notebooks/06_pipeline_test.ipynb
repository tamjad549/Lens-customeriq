```python
import sys
sys.path.append("..")

import joblib
import pandas as pd
from src import features as F

raw = pd.read_csv("../data/raw/online_retail_II.csv")
clean = F.clean_transactions(raw)
print("Clean rows:", len(clean))          # 777912 aana chahiye

feat = F.build_features(clean)
scaler = joblib.load("../models/rfm_scaler.joblib")
km = joblib.load("../models/kmeans_rfm.joblib")
model = joblib.load("../models/churn_xgb.joblib")

result = F.predict_churn(F.assign_segments(feat, scaler, km), model)
print(result.shape)
result.sort_values("ChurnRisk", ascending=False).head()
```
